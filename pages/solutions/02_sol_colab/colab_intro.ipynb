```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import Video

fig, ax = plt.subplots()

x = []
y = []

line, = ax.plot([], [])

ax.set_xlim(0, 10)
ax.set_ylim(-1, 1)

def update(frame):
    x.append(frame / 10)
    y.append(np.sin(frame / 10))
    line.set_data(x, y)
    return line,

ani = FuncAnimation(fig, update, frames=100)

ani.save("video.mp4", writer="ffmpeg")

Video("video.mp4", embed=True)
```

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 20*np.pi, 1000)

x = np.cos(3*t)
y = np.sin(2*t)

plt.figure(figsize=(6,6))
plt.plot(x, y)

plt.title("Krzywa parametryczna")
plt.savefig("parametric_curve.png")
plt.show()
```
